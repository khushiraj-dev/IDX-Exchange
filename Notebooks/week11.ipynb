```python
import pandas as pd
from pathlib import Path

folder = Path(r"C:\Users\khush\idx files")
sold = pd.read_csv(folder / "sold_combined.csv", low_memory=False)
listings = pd.read_csv(folder / "listings_combined.csv", low_memory=False)

# Filter to San Ramon
sr_sold = sold[sold['City'] == 'San Ramon'].copy()
sr_listings = listings[listings['City'] == 'San Ramon'].copy()

# Convert dates
sr_sold['CloseDate'] = pd.to_datetime(sr_sold['CloseDate'], errors='coerce')
sr_sold['ListingContractDate'] = pd.to_datetime(sr_sold['ListingContractDate'], errors='coerce')

# Market Overview
print("=== MARKET OVERVIEW ===")
print(f"Total sold transactions: {len(sr_sold):,}")
print(f"Median close price: ${sr_sold['ClosePrice'].median():,.0f}")
print(f"Average close price: ${sr_sold['ClosePrice'].mean():,.0f}")
print(f"Median days on market: {sr_sold['DaysOnMarket'].median():.0f} days")

# Pricing Trends
print("\n=== PRICING TRENDS ===")
sr_sold['price_per_sqft'] = sr_sold['ClosePrice'] / sr_sold['LivingArea']
sr_sold['price_ratio'] = sr_sold['ClosePrice'] / sr_sold['OriginalListPrice']
print(f"Median price per sqft: ${sr_sold['price_per_sqft'].median():,.0f}")
print(f"Average sale-to-list ratio: {sr_sold['price_ratio'].median():.3f}")
print(f"% sold above list price: {(sr_sold['price_ratio'] > 1).mean()*100:.1f}%")

# Market Activity
print("\n=== MARKET ACTIVITY ===")
print(f"Total new listings: {len(sr_listings):,}")
print(f"Total closed sales: {len(sr_sold):,}")

# Competitive Landscape
print("\n=== TOP 5 LISTING AGENTS ===")
top_agents = sr_sold.groupby('ListAgentFullName').agg(
    units=('ClosePrice', 'count'),
    volume=('ClosePrice', 'sum'),
    median_price=('ClosePrice', 'median')
).sort_values('units', ascending=False).head(5)
print(top_agents)

print("\n=== TOP 5 LISTING OFFICES ===")
top_offices = sr_sold.groupby('ListOfficeName').agg(
    units=('ClosePrice', 'count'),
    volume=('ClosePrice', 'sum'),
    median_price=('ClosePrice', 'median')
).sort_values('units', ascending=False).head(5)
print(top_offices)

# Year over year trend
print("\n=== YEARLY TREND ===")
sr_sold['Year'] = sr_sold['CloseDate'].dt.year
yearly = sr_sold.groupby('Year')['ClosePrice'].median()
print(yearly)
```

```python
# Filter to Fremont
fremont_sold = sold[sold['City'] == 'Fremont'].copy()
fremont_listings = listings[listings['City'] == 'Fremont'].copy()

# Convert dates
fremont_sold['CloseDate'] = pd.to_datetime(fremont_sold['CloseDate'], errors='coerce')

# Market Overview
print("=== MARKET OVERVIEW ===")
print(f"Total sold transactions: {len(fremont_sold):,}")
print(f"Median close price: ${fremont_sold['ClosePrice'].median():,.0f}")
print(f"Average close price: ${fremont_sold['ClosePrice'].mean():,.0f}")
print(f"Median days on market: {fremont_sold['DaysOnMarket'].median():.0f} days")

# Pricing Trends
print("\n=== PRICING TRENDS ===")
fremont_sold['price_per_sqft'] = fremont_sold['ClosePrice'] / fremont_sold['LivingArea']
fremont_sold['price_ratio'] = fremont_sold['ClosePrice'] / fremont_sold['OriginalListPrice']
print(f"Median price per sqft: ${fremont_sold['price_per_sqft'].median():,.0f}")
print(f"Median sale-to-list ratio: {fremont_sold['price_ratio'].median():.3f}")
print(f"% sold above list price: {(fremont_sold['price_ratio'] > 1).mean()*100:.1f}%")

# Market Activity
print("\n=== MARKET ACTIVITY ===")
print(f"Total new listings: {len(fremont_listings):,}")
print(f"Total closed sales: {len(fremont_sold):,}")

# Top agents
print("\n=== TOP 5 LISTING AGENTS ===")
top_agents = fremont_sold.groupby('ListAgentFullName').agg(
    units=('ClosePrice', 'count'),
    volume=('ClosePrice', 'sum'),
    median_price=('ClosePrice', 'median')
).sort_values('units', ascending=False).head(5)
print(top_agents)

# Top offices
print("\n=== TOP 5 LISTING OFFICES ===")
top_offices = fremont_sold.groupby('ListOfficeName').agg(
    units=('ClosePrice', 'count'),
    volume=('ClosePrice', 'sum'),
    median_price=('ClosePrice', 'median')
).sort_values('units', ascending=False).head(5)
print(top_offices)

# Yearly trend
print("\n=== YEARLY TREND ===")
fremont_sold['Year'] = fremont_sold['CloseDate'].dt.year
yearly = fremont_sold.groupby('Year')['ClosePrice'].median()
print(yearly)
```

```python
# Buyer's vs Seller's Market Analysis
print("=== MARKET TYPE ANALYSIS ===")

# Months of supply = (active listings / monthly sales)
monthly_sales = len(fremont_sold) / 30  # 30 months of data
months_of_supply = len(fremont_listings) / monthly_sales
print(f"Monthly sales rate: {monthly_sales:.0f} homes/month")
print(f"Months of supply: {months_of_supply:.1f} months")

if months_of_supply < 3:
    market_type = "Strong Seller's Market"
elif months_of_supply < 6:
    market_type = "Balanced Market"
else:
    market_type = "Buyer's Market"

print(f"Market classification: {market_type}")
print(f"Supporting indicators:")
print(f"  - Median DOM: {fremont_sold['DaysOnMarket'].median():.0f} days (< 30 = seller's)")
print(f"  - Sale-to-list ratio: {fremont_sold['price_ratio'].median():.3f} (> 1.0 = seller's)")
print(f"  - % above list: {(fremont_sold['price_ratio'] > 1).mean()*100:.1f}% (> 50% = seller's)")
```

```python
# Yearly trend analysis
print("=== YEARLY PRICE TREND ===")
fremont_sold['Year'] = fremont_sold['CloseDate'].dt.year
yearly = fremont_sold.groupby('Year').agg(
    median_price=('ClosePrice', 'median'),
    total_sales=('ClosePrice', 'count'),
    median_dom=('DaysOnMarket', 'median'),
    pct_above_list=('price_ratio', lambda x: (x > 1).mean() * 100)
).round(2)
print(yearly)

# Property subtype breakdown
print("\n=== SALES BY PROPERTY TYPE ===")
subtype = fremont_sold.groupby('PropertySubType').agg(
    units=('ClosePrice', 'count'),
    median_price=('ClosePrice', 'median'),
    median_dom=('DaysOnMarket', 'median')
).sort_values('units', ascending=False)
print(subtype)

# Price distribution
print("\n=== PRICE DISTRIBUTION ===")
bins = [0, 1000000, 1500000, 2000000, 2500000, float('inf')]
labels = ['Under $1M', '$1M-$1.5M', '$1.5M-$2M', '$2M-$2.5M', 'Over $2.5M']
fremont_sold['price_range'] = pd.cut(fremont_sold['ClosePrice'], bins=bins, labels=labels)
print(fremont_sold['price_range'].value_counts().sort_index())
```
